Where should the labels come from?
Every one of the shop’s projects needs labeled data, and there are four main ways you might think about getting labels, like we talked about in class:
Historical data: records of what actually happened, such as the shop’s repair records for vehicles it sold in past years.
Human annotation: people (e.g., shop’s mechanics or staff) look at each example and label it.
Generated labels: an AI model, such as an LLM, reads each example and outputs a label. These are then used to train another model.
Weak supervision: simple rules or heuristics are combined to produce labels automatically.
Below are five of the shop’s tasks. For each one, choose the label source you think fits best (you can combine sources if you explain how), say why in a sentence or two, and name the biggest risk of that choice for this task.

    Label Source - Historical Data
    
    Why - It says the shop has kept repair records for vehicles sold in the past years which is exactly the data they would need to answer their question

    Risk - Older data may not account for newer vehicles cost of repair. Inflation and other factors can directly impact costs as well.

Rust: does an inspection photo show rust damage?

    Label Source - Human Annotation

    Why - Rust can be a very ambiguous concept visually. The presence of rust in a photograph is not subject to a set of rules or data from previous events.

    Risk - One individual to another may have a different opinion on whether there is Rust or not in the photograph.

Misleading listings: is a seller’s listing description misleading about the vehicle’s condition?

    Label Source - Human Annotation

    Why - It would be very difficult for any other source other than a human to detect deception. Rules, Data, or AI would take input at face value. Humans are more apt to take in data while keeping Human behaviors such as deception in context.

    Risk - Human context and intuition can be shaky, people may have different feelings about the label given different circumstances of their current state. Also, as always, different annotators may have different interpretations of the vehicles condition relative to it's description

You could have it track a couple as to the logic behind it's decision, but you would never be able to comprehend how it would make any decision given it's extreme complexity in it's decision making process.

Search tags: which features (lift kit, roof rack, winch, and so on) does a listing mention, for the shop’s website search?

    Label Source - Weak Supervision
